```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('/content/Top 500 Movies Ranked by Combined Critics and Audience Scores.csv')
```

```python
df.sample(5)
```

```python
df.describe()
```

```python
df.info()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.scatter(df['IMDb_10'],df['Critic_Reviews'])
plt.xlabel('IMDb_10')
plt.ylabel('Critic_Reviews')
plt.title('IMDb_10 vs Critic_Reviews')
plt.show()
```

```python
sns.pairplot(df)
```

```python
sns.kdeplot(df['IMDb_100'])
```

```python
sns.lineplot(df['Year'].value_counts())
```

```python
df['Year'].value_counts()
```

```python
df['Streaming_On'].str.replace('itunes','')
```

```python
df['Streaming_On'].str.replace('itunes,','').value_counts().head(10).plot(kind='pie',autopct='%1.1f%%',figsize=(10,10))
```

```python
missing=df.isnull().sum().sort_values(ascending=False)
missing[missing>0].plot(kind='bar')
plt.title('Missing Values')
plt.show()
```

```python
sns.histplot(df['Year'],bins=30)
plt.title('movies by year')
```

```python
genre_counts = (
    df['Genre']
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
    .head(15)
)

plt.figure(figsize=(10,6))
sns.barplot(x=genre_counts.values,
            y=genre_counts.index)

plt.title("Top Genres")
plt.show()
```

```python
sns.histplot(df['Audience_Rating'],bins=25,kde=True)
```

```python
sns.boxplot(x=df['Google_Score'])
plt.title('Google score distribution')
```

```python
num_cols = df.select_dtypes(include=np.number)
plt.figure(figsize=(10,8))
sns.heatmap(
    num_cols.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.show()
```

```python
top_directors=df['Director'].value_counts().head(15)
```

```python
sns.barplot(y=top_directors.index,x=top_directors.values)
plt.title('Top Directors')
plt.show()
```

```python
top_languages=df['Language'].value_counts().head(10)
```

```python
sns.barplot(y=top_languages.index,x=top_languages.values)
plt.title('Top Languages')
plt.show()
```

```python
platforms = (
    df['Streaming_On']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
    .head(15)
)

plt.figure(figsize=(10,6))

sns.barplot(
    x=platforms.values,
    y=platforms.index
)

plt.title("Most Popular Streaming Platforms")
plt.show()
```
