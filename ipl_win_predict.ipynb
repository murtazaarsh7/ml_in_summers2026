```python
df=df[df['innings']==2]
```

```python
df['batting_team'].unique()
```

```python
teams=['Kolkata Knight Riders', 'Rajasthan Royals',
       'Mumbai Indians', 'Chennai Super Kings'
       ,'Sunrisers Hyderabad','Delhi Capitals', 'Punjab Kings', 'Lucknow Super Giants',
       'Royal Challengers Bengaluru','Gujarat Titans']
```

```python
df['batting_team']=df['batting_team'].str.replace('Delhi Daredevils','Delhi Capitals')
df['bowling_team']=df['bowling_team'].str.replace('Delhi Daredevils','Delhi Capitals')

df['batting_team']=df['batting_team'].str.replace('Royal Challengers Bangalore','Royal Challengers Bangalore')
df['bowling_team']=df['bowling_team'].str.replace('Royal Challengers Bangalore','Royal Challengers Bengaluru')

df['batting_team']=df['batting_team'].str.replace('Kings XI Punjab','Punjab Kings')
df['bowling_team']=df['bowling_team'].str.replace('Kings XI Punjab','Punjab Kings')

```

```python
df=df[df['batting_team'].isin(teams)]
df=df[df['bowling_team'].isin(teams)]
```

```python
df.shape
```

```python
df.info()
```

```python
df.drop_duplicates(inplace=True)
```

```python
new_df=df[['match_id','batting_team','bowling_team','city','over','ball','team_wicket','runs_total_x','runs_total_y','team_wicket','batting_partners','team_runs','innings','match_won_by']]
```

```python
new_df
```

```python
new_df['current_score']=new_df.groupby('match_id')['runs_total_x'].cumsum()
```

```python
new_df['runs_left']=new_df['runs_total_y']-new_df['current_score']+1
```

```python
new_df['balls_left']=120-(new_df['over']*6+new_df['ball'])
```

```python
new_df
```

```python
new_df['wickets_left']=10-df['team_wicket']
```

```python
new_df
```

```python
new_df['crr']=new_df['current_score']*6/(120-new_df['balls_left'])
```

```python
new_df
```

```python
new_df['rrr']=(new_df['runs_left']*6)/new_df['balls_left']
```

```python
new_df
```

```python
def result(row):
  return 1 if row['batting_team']==row['match_won_by'] else 0
```

```python
new_df['result']=new_df.apply(result,axis=1)
```

```python
new_df=new_df[['batting_team','bowling_team','city','runs_left','balls_left','wickets_left','runs_total_y','crr','rrr','result']]
```

```python
new_df
```

```python
new_df=new_df.sample(new_df.shape[0])
```

```python
new_df.sample()
```

```python
new_df.isnull().sum()
```

```python
new_df=new_df[new_df['balls_left']!=0]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x=new_df.iloc[:,:-1]
y=new_df.iloc[:,-1]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=1)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python

trf=ColumnTransformer([
    ('trf',OneHotEncoder(sparse_output=False,drop='first'),['batting_team','bowling_team','city'])
],remainder='passthrough')
```

```python

trf
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
```

```python

pipe=Pipeline(steps=[
    ('step1',trf),
    ('step2',LogisticRegression(solver='liblinear'))]
)
```

```python
pipe.fit(x_train,y_train)
```

```python
from sklearn.metrics import accuracy_score
```

```python
y_pred=pipe.predict(x_test)
```

```python
accuracy_score(y_test,y_pred)
```
